```python
# ==============================
# Initialize NLP Components
# ==============================

stop_words = set(stopwords.words('english'))

lemmatizer = WordNetLemmatizer()
```

```python
# ==============================
# Text Preprocessing Function
# ==============================

def preprocess_text(text):

    # Convert to lowercase
    text = text.lower()

    # Remove numbers
    text = re.sub(r'\d+', '', text)

    # Remove punctuation
    text = text.translate(str.maketrans('', '', string.punctuation))

    # Remove extra spaces
    text = re.sub(r'\s+', ' ', text).strip()

    # Tokenization
    tokens = word_tokenize(text)

    # Remove stopwords
    tokens = [
        word for word in tokens
        if word not in stop_words
    ]

    # Lemmatization
    tokens = [
        lemmatizer.lemmatize(word)
        for word in tokens
    ]

    # Join tokens
    cleaned_text = " ".join(tokens)

    return cleaned_text
```

```python
df['cleaned_text'] = df['text'].apply(preprocess_text)

df[['text', 'cleaned_text']].head()
```

```python
# ==============================
# Save Preprocessed Dataset
# ==============================

output_file = "preprocessed_dataset.csv"

df.to_csv(output_file, index=False)

print(f"Saved as: {output_file}")
```

```python
# ==============================
# Create Tokenized Corpus
# ==============================

corpus = df["cleaned_text"].apply(lambda x: x.split()).tolist()

print("Total Documents:", len(corpus))

print("\nFirst Document Tokens:\n")

print(corpus[0][:30])
```

```python
import numpy as np
import pandas as pd

from gensim.models import Word2Vec
```

```python
# ==============================
# Train Word2Vec Model
# ==============================

word2vec_model = Word2Vec(
    sentences=corpus,    # Tokenized documents used for training
    vector_size=100,     # Size of each word vector (100 numerical values)
    window=5,            # Number of nearby/context words to consider
    min_count=2,         # Ignore words appearing fewer than 2 times
    workers=4,           # Number of CPU threads used for training
    sg=1,                # 1 = Skip-gram | 0 = CBOW
    epochs=20,           # Number of complete training iterations
    seed=42              # Ensures reproducible results
)

print("Word2Vec Model Training Completed")
```

```python
# ==============================
# Model Information
# ==============================

print("Vocabulary Size :", len(word2vec_model.wv))

print("Embedding Dimension :", word2vec_model.vector_size)
```

```python
# ============================================
# Most Similar Words
# ============================================

word = input("Enter a word: ").lower()

if word in word2vec_model.wv:

    similar_words = word2vec_model.wv.most_similar(word, topn=10)

    print("\nMost Similar Words\n")

    for w, score in similar_words:
        print(f"{w:20} {score:.4f}")

else:

    print("Word not found in vocabulary.")
```

```python
# ============================================
# Save Word2Vec Model
# ============================================

word2vec_model.save("word2vec_semantic.model")

print("Model Saved Successfully")
```

```python
# ================================================
# Document Embedding Function
# ================================================

def document_vector(text, model):

    words = text.split()

    word_vectors = [
        model.wv[word]
        for word in words
        if word in model.wv
    ]

    if len(word_vectors) == 0:
        return np.zeros(model.vector_size)

    return np.mean(word_vectors, axis=0)
```

```python
# ================================================
# Generate Document Embeddings
# ================================================

document_embeddings = np.array(
    df["cleaned_text"].apply(
        lambda x: document_vector(x, word2vec_model)
    ).tolist()
)

print("Document Embeddings Created")

print("Shape :", document_embeddings.shape)
```

```python
# ================================================
# Query Preprocessing,User Query
# ================================================

def preprocess_query(query):

    return preprocess_text(query)
```

```python
# ================================================
# Query Embedding
# ================================================

def query_vector(query):

    cleaned_query = preprocess_query(query)

    return document_vector(cleaned_query, word2vec_model)
```

```python
# ================================================
# Semantic Search Function
# ================================================
from sklearn.metrics.pairwise import cosine_similarity
def semantic_search(query, top_k=5):

    query_embedding = query_vector(query)

    similarity_scores = cosine_similarity(
        [query_embedding],
        document_embeddings
    )[0]

    top_indices = np.argsort(similarity_scores)[::-1][:top_k]

    results = df.iloc[top_indices].copy()

    results["Similarity Score"] = similarity_scores[top_indices]

    return results[
        [
            "document_id",
            "title",
            "category",
            "Similarity Score"
        ]
    ]
```

```python
# ================================================
# Test Semantic Search
# ================================================

query = input("Enter your search query : ")

results = semantic_search(query, top_k=5)

results
```

```python
fasttext_model.save("fasttext.model")

print("Model Saved Successfully")
```

```python
def generate_text_embedding(tokens, model):

    vectors = []

    for word in tokens:

        vectors.append(model.wv[word])

    return np.mean(vectors, axis=0)
```

```python
# Document Embedding
document_embeddings = []

for document in corpus:

    embedding = generate_text_embedding(
        document,
        fasttext_model
    )

    document_embeddings.append(embedding)

document_embeddings = np.array(document_embeddings)

print(document_embeddings.shape)
```

```python
np.save(
    "fasttext_document_embeddings.npy",
    document_embeddings
)

print("Document Embeddings Saved Successfully")
```

Comparison with TfidfVectorizer

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity


vectorizer = TfidfVectorizer()

tfidf_matrix = vectorizer.fit_transform(df["cleaned_text"])

print(tfidf_matrix.shape)
```

```python
def tfidf_search(query, top_k=5):

    query_vector = vectorizer.transform([query])

    similarity_scores = cosine_similarity(
        query_vector,
        tfidf_matrix
    ).flatten()

    results = df.copy()

    results["Similarity_Score"] = similarity_scores

    results = results.sort_values(
        by="Similarity_Score",
        ascending=False
    )

    return results[
        [
            "document_id",
            "category",
            "title",
            "Similarity_Score"
        ]
    ].head(top_k)
```

```python
tfidf_results = tfidf_search(
    "how to check the order receipt",
    top_k=5
)

tfidf_results
```

```python
word2vec_results = semantic_search(
    "how to check the order receipt",
    top_k=5
)

word2vec_results
```
